```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv(
    "motorcycle_mileage_dataset.csv",
    low_memory=False
)
```

```python
print("Shape:", data.shape)

print("\nFirst 5 rows:")
print(data.head())

print("\nDataset information:")
data.info()

print("\nStatistical summary:")
print(data.describe())
```

```python
missing_values = data.isnull().sum()

print(missing_values[missing_values > 0])
```

```python
print("Exact duplicate rows:", data.duplicated().sum())

print(
    "Duplicate record IDs:",
    data["record_id"].duplicated().sum()
)
```

```python
print(data["mileage_kmpl"].describe())

plt.figure(figsize=(10, 5))

sns.histplot(
    data=data,
    x="mileage_kmpl",
    bins=50
)

plt.title("Motorcycle Mileage Distribution")
plt.xlabel("Mileage (km/l)")
plt.ylabel("Frequency")

plt.show()
```
